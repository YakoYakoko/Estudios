```python

def beam_search(graph, start, goal, beam_width):
    # El haz inicial contiene: (nodo_actual, ruta_recorrida, puntaje_acumulado)
    beam = [(start, [start], 0)]

    while beam:
        candidates = []

        # 1. Expandir los nodos del haz actual
        for node, path, score in beam:
            if node == goal:
                return path, score

            for neighbor, edge_score in graph.get(node, []):
                new_score = score + edge_score
                candidates.append(
                    (neighbor, path + [neighbor], new_score)
                )

        if not candidates:
            break

        # 2. Ordenar candidatos por puntaje de forma descendente y aplicar poda (K = beam_width)
        candidates.sort(key=lambda x: x[2], reverse=True)
        beam = candidates[:beam_width]

        # 3. Verificar si la meta se alcanzó dentro de los mejores K candidatos seleccionados
        for node, path, score in beam:
            if node == goal:
                return path, score

    return None

graph = {
    'A': [('B', 8), ('C', 6), ('D', 4)],
    'B': [('E', 7), ('F', 5)],
    'C': [('G', 9), ('H', 3)],
    'D': [('I', 6)],
    'E': [],
    'F': [],
    'G': [],
    'H': [],
    'I': []
}

result = beam_search(graph, 'A', 'G', 2)

print("Mejor Ruta Encontrada:", result[0])
print("Puntaje Acumulado:", result[1])
```

```python
import math


vocab = {1: "El", 2: "gato", 3: "perro", 4: "corre", 5: "duerme", 6: "<eos>"}

def mock_llm_logits(sequence):
    last_token = sequence[-1] if sequence else 1


    if last_token == 1:
        return {2: -0.2, 3: -0.8}
    elif last_token == 2:
        return {4: -0.3, 5: -0.6, 6: -2.5}
    elif last_token == 3:
        return {4: -0.5, 5: -0.4, 6: -2.0}
    else:
        return {6: -0.1}

def llm_text_beam_search(start_tokens, beam_width, max_steps):

    beam = [(start_tokens, 0.0)]

    for _ in range(max_steps):
        candidates = []

        for tokens, score in beam:

            if tokens[-1] == 6:
                candidates.append((tokens, score))
                continue


            next_token_probs = mock_llm_logits(tokens)

            for next_token, log_prob in next_token_probs.items():
                new_tokens = tokens + [next_token]
                new_score = score + log_prob
                candidates.append((new_tokens, new_score))


        candidates.sort(key=lambda x: x[1], reverse=True)


        beam = candidates[:beam_width]

    return beam


resultado = llm_text_beam_search(start_tokens=[1], beam_width=2, max_steps=3)

print("=== RESULTADOS DE GENERACIÓN DE TEXTO CON BEAM SEARCH ===")
for tokens, score in resultado:
    texto_generado = " ".join([vocab[t] for t in tokens if t not in [6]])
    print(f"Secuencia de Tokens: {tokens}")
    print(f"Texto decodificado: '{texto_generado}' | Log-Probabilidad: {score:.4f}\n")
```
